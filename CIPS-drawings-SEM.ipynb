```python
import pandas as pd
from semopy import Model

model_spec = """
    # Structural model
    confused ~ male + cips
    self ~ male + cips
    product ~ male + cips
	cips ~ male
"""
# read in data using pandas
data = pd.read_excel("https://docs.google.com/spreadsheets/d/1RJPBLFbfPwA_LET_17PjXecvpsT6po89q8z8UTyY_yk/export",engine='openpyxl',index_col=0)
# data = pd.read_csv("my_data_file.csv", index_col=0)

m = Model(model_spec)
r = m.fit(data)
print(r)
ins = m.inspect()
ins # This will show the betas, their std errors, and p-values. Use for the two elements, self and confusion.
```
